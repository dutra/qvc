```python
from speclite.filters import FilterResponse
from speclite import filters

def populate_lam_flux(df_agn_all):
    def load_spec_from_cache(sdss_name, cache_dir="data/spectra_cache"):
        """Return cached FITS HDUList if available, else None."""
        cache_file = os.path.join(cache_dir, f"{sdss_name}.fits")
        if os.path.exists(cache_file):
            return fits.open(cache_file, memmap=False)
        return None

    bands = ['u', 'g', 'r', 'i', 'z']
    sdss_filters = filters.load_filters(*[f'sdss2010-{b}' for b in bands])

    # Iterate over each row in df_agn and populate flux using sdss_name
    flux_data = []
    fiberMag_data = {b: [] for b in bands}
    lam_min = []
    lam_max = []
    lam_range = []
    
    # Iterate over each row in df_agn with a progress bar
    for _, row in tqdm(df_agn_all.iterrows(), total=len(df_agn_all), desc="Processing spectra"):
        sdss_name = row["sdss_name"]
        hdul = load_spec_from_cache(sdss_name)
        if hdul is not None:
            lam = 10**hdul[1].data['loglam']  # Extract wavelength
            flux = hdul[1].data['flux']  # Extract flux
            
            for b, filt in zip(bands, sdss_filters):
                try:
                    mag_synth = filt.get_ab_magnitude(
                                1e-17 * flux * u.erg / u.s / u.cm**2 / u.AA,
                                lam * u.AA
                        )
                    
                except Exception as e:
                    #print(f"Error computing synthetic magnitude for {sdss_name} in {b}-band: {e}")
                    mag_synth = np.nan
                fiberMag_data[b].append(mag_synth)
            lam_rf = lam / (1 + row['z'])
            flux_data.append(np.nanmean(flux))
            lam_min.append(np.nanmin(lam_rf))
            lam_max.append(np.nanmax(lam_rf))
            lam_range.append(np.nanmax(lam_rf) - np.nanmin(lam_rf))
        else:
            for band in bands:
                fiberMag_data[band].append(None)  # Append None if spectrum is not available
            flux_data.append(None)  # Append None if spectrum is not available
            lam_min.append(None)
            lam_max.append(None)
            lam_range.append(None)

    # Add the flux data as a new column in df_agn
    df_agn_all["flux"] = flux_data

    for b in bands:
        df_agn_all[f"fiberMag_{b}"] = fiberMag_data[b]
    df_agn_all["lam_min"] = lam_min 
    df_agn_all["lam_max"] = lam_max 
    df_agn_all["lam_range"] = lam_range

    # df_agn_all['PL_break_wave_inbounds'] = df_agn_all['PL_break_wave'].between(df_agn_all['lam_min'], df_agn_all['lam_max'])
    # df_agn_all['PL_break_wave_inbounds'].value_counts()
    return df_agn_all

from types import SimpleNamespace


df_agn, df_agn_all = load_agn_data("results/data/oct26a_carma_single_lagblrband_chisq_spl_nofhost_bwb_taufasttruncated_lmc-6_N1w1000s200t14ch4.h5",
                       fhost_cut=0.0,
                       only_load=False,
                       populate_sdss=False,
                       spectra_fit_csv=['results/data/nov1c_collect_scratch_oct27b_carma_nofhost_iron_best.csv'],
                       zquery_csv='results/data/sep19_chisq_zquery.csv',
                       args=SimpleNamespace(wrms_cut=2))


#df_agn_all = populate_lam_flux(df_agn_all)
#df_agn = populate_lam_flux(df_agn)
```

```python
# --- Helper: crisp isocontour where log_lbol == threshold (uses df_plot['log_lbol']) ---
def draw_lbol_isocontour(
    ax, df_plot, *,
    threshold=45,
    x='z', y='apparent_mag_2500',
    nbins=30,
    min_pts=20,
    min_mag_span=0.2,
    slope_eps=1e-4,
    smooth_window=0,
    line_kw=None,
    return_points=False,
):
    """
    Draw the y(x) curve where log_lbol == threshold by locally regressing
    log_lbol ~ a * y + b (within z-bins) and solving y* = (threshold - b)/a.
    """
    import numpy as np
    import pandas as pd

    line_kw = dict(lw=2.2, ls='-', color='black', alpha=0.95, zorder=5) if line_kw is None else line_kw

    # Clean arrays
    z = pd.to_numeric(df_plot[x], errors='coerce').to_numpy()
    m = pd.to_numeric(df_plot[y], errors='coerce').to_numpy()
    l = pd.to_numeric(df_plot['log_lbol'], errors='coerce').to_numpy()
    ok = np.isfinite(z) & np.isfinite(m) & np.isfinite(l)
    if not ok.any():
        return () if return_points else None
    z, m, l = z[ok], m[ok], l[ok]

    # Bin in z
    zmin, zmax = float(np.nanmin(z)), float(np.nanmax(z))
    if not np.isfinite(zmin) or not np.isfinite(zmax) or zmax <= zmin:
        return () if return_points else None

    edges   = np.linspace(zmin, zmax, nbins + 1)
    centers = 0.5 * (edges[:-1] + edges[1:])
    m_star  = np.full(nbins, np.nan, dtype=float)

    # Per-bin weighted linear fit: l ~ a*m + b (emphasize near threshold)
    for i in range(nbins):
        in_bin = (z >= edges[i]) & (z < edges[i+1])
        if in_bin.sum() < min_pts:
            continue

        m_i = m[in_bin]
        l_i = l[in_bin]
        if np.nanmax(m_i) - np.nanmin(m_i) < min_mag_span:
            continue

        eps = 1e-6
        w = 1.0 / (np.abs(l_i - threshold) + eps)

        X = np.vstack([m_i, np.ones_like(m_i)]).T
        XT_W = (X.T * w) @ X
        bvec = (X.T * w) @ l_i
        try:
            a_hat, b_hat = np.linalg.solve(XT_W, bvec)
        except np.linalg.LinAlgError:
            a_hat, b_hat = np.polyfit(m_i, l_i, 1)

        if not np.isfinite(a_hat) or abs(a_hat) < slope_eps:
            continue

        m_cut = (threshold - b_hat) / a_hat
        m_lo, m_hi = np.nanpercentile(m_i, [2, 98])
        if not (m_lo - 1.0 <= m_cut <= m_hi + 1.0):
            continue

        m_star[i] = m_cut

    # Safe smoothing across bins
    import numpy as np
    if smooth_window and smooth_window > 1 and np.isfinite(m_star).sum() >= 2:
        k = int(smooth_window)
        if k % 2 == 0:
            k += 1
        # Fill NaNs by interpolation with edge clamping
        y_fill = m_star.copy()
        idx = np.arange(y_fill.size)
        good = np.isfinite(y_fill)
        if good.any():
            first = idx[good][0]; last = idx[good][-1]
            y_fill[:first] = y_fill[first]
            y_fill[last+1:] = y_fill[last]
            y_fill = np.interp(idx, idx[good], y_fill[good])
        pad = k // 2
        y_pad = np.pad(y_fill, (pad, pad), mode='edge')
        kernel = np.ones(k) / k
        m_star_sm = np.convolve(y_pad, kernel, mode='valid')
    else:
        m_star_sm = m_star

    good = np.isfinite(m_star_sm) & np.isfinite(centers)
    if good.sum() >= 2:
        ax.plot(centers[good], m_star_sm[good], **line_kw)

    return (centers[good], m_star_sm[good]) if return_points else None


# ---- Helper: robust colorbar with label (works for continuous/categorical) ----
def add_colorbar(ax, mappable, label, *, fraction=0.046, pad=0.04, ticks=None, ticklabels=None):
    import numpy as np
    fig = ax.figure
    cb = fig.colorbar(mappable, ax=ax, fraction=fraction, pad=pad)
    if ticks is not None:
        cb.set_ticks(ticks)
    if ticklabels is not None:
        cb.set_ticklabels(ticklabels)
    cb.set_label(label)
    cb.ax.tick_params(length=3)
    return cb


# ---- Helper: build categorical mappable (codes, cmap, norm, mappable, ticks, labels) ----
def categorical_mappable(categories, colors):
    """
    categories: list of category labels in the order you want them to appear
    colors: same length list of color specs
    Returns: (norm, cmap, mappable, ticks, ticklabels)
    """
    import numpy as np
    import matplotlib as mpl
    from matplotlib.colors import ListedColormap, BoundaryNorm

    n = len(categories)
    cmap = ListedColormap(colors, name='catmap')
    bounds = np.arange(-0.5, n + 0.5, 1.0)
    norm = BoundaryNorm(bounds, cmap.N)
    mappable = mpl.cm.ScalarMappable(norm=norm, cmap=cmap)
    mappable.set_array([])  # dummy
    ticks = np.arange(n)
    ticklabels = [str(c) for c in categories]
    return norm, cmap, mappable, ticks, ticklabels
# ======================== PLOTTING ========================
cmap_chi  = plt.get_cmap('viridis')
cmap_lbol = plt.get_cmap('viridis')
cmap_host = plt.get_cmap('viridis')
cmap_conti = plt.get_cmap('viridis')
cmap_alpha = plt.get_cmap('viridis')
cmap_sn_median = plt.get_cmap('viridis')
cmap_lam = plt.get_cmap('viridis')

ncols = 17
fig, axes = plt.subplots(2, ncols, figsize=(6*ncols, 10), sharey=True, sharex='col', constrained_layout=True)

#fhost_cut = 0.1
#df_cut = df_agn_all[df_agn_all['f_host_2500'] < fhost_cut]
dfs = [df_agn_all, df_agn]
titles = ['All', 'Final selection']

for i, (df, title) in enumerate(zip(dfs, titles)):
    df_plot = df.copy()
    df_plot = df_plot[df_plot['best'] == True]
    #df_plot = df_plot[(df_plot['apparent_mag_2500'].between(0, 30)) & (df_plot['log_lbol'] > 10)]
    print(len(df_plot), "points in", title)

    # Panel 0: delta_m_avg (continuous; new first panel)
    delta_m_avg = pd.to_numeric(df_plot['delta_m_avg'], errors='coerce')
    vmin, vmax = np.percentile(delta_m_avg.dropna(), [5, 95])
    sc = axes[i, 0].scatter(
        df_plot['z'], df_plot['apparent_mag_2500'],
        c=np.clip(delta_m_avg, vmin, vmax),
        cmap='coolwarm', s=8, alpha=0.6, vmin=vmin, vmax=vmax
    )
    add_colorbar(axes[i, 0], sc, f'delta_m_avg (clipped {vmin:.2g}–{vmax:.2g})')
    draw_lbol_isocontour(axes[i, 0], df_plot)
    axes[i, 0].set_title(f'Color: delta_m_avg ({title})')
    axes[i, 0].set_xlabel('Redshift (z)')
    axes[i, 0].set_ylabel('apparent_mag_2500')


    # Panel 1: redchi2_conti_full (continuous)
    chi = pd.to_numeric(df_plot['redchi2_conti_full']-1, errors='coerce')
    vmin_c, vmax_c = 1, 2
    sc_chi = axes[i, 1].scatter(
        df_plot['z'], df_plot['apparent_mag_2500'],
        c=np.clip(chi, vmin_c, vmax_c),
        cmap=cmap_chi, s=8, alpha=0.3, vmin=vmin_c, vmax=vmax_c
    )
    add_colorbar(axes[i, 1], sc_chi, f'redchi2_conti_full (clipped {vmin_c:.2g}–{vmax_c:.2g})')
    draw_lbol_isocontour(axes[i, 1], df_plot)
    axes[i, 1].set_title(f'Color: redchi2_conti_full ({title})')
    axes[i, 1].set_xlabel('Redshift (z)')

    # Panel 2: failed (categorical)
    failed_mask = (df_plot['decomp_host'] == True) & (df_plot['f_host_2500'] < 1e-3)
    cat_vals = np.where(failed_mask, 1, 0)
    cats = ['other', 'failed']
    cols = ['black', 'red']
    norm, cat_cmap, mappable, ticks, ticklabels = categorical_mappable(cats, cols)
    sc_failed = axes[i, 2].scatter(
        df_plot['z'], df_plot['apparent_mag_2500'],
        c=cat_vals, cmap=cat_cmap, norm=norm, s=8, alpha=0.5
    )
    add_colorbar(axes[i, 2], mappable, 'status', ticks=ticks, ticklabels=ticklabels)
    draw_lbol_isocontour(axes[i, 2], df_plot)
    axes[i, 2].set_title(f'Failed (decomp_host & f_host_2500 < 1e-3) ({title})')
    axes[i, 2].set_xlabel('Redshift (z)')

    # Panel 3: conti_a_0 (continuous)
    sc_conti = axes[i, 3].scatter(
        df_plot['z'], df_plot['apparent_mag_2500'],
        c=df_plot['conti_a_0'], cmap=cmap_conti, s=8, alpha=0.5
    )
    add_colorbar(axes[i, 3], sc_conti, 'conti_a_0')
    draw_lbol_isocontour(axes[i, 3], df_plot)
    axes[i, 3].set_title(f'Color: conti_a_0 ({title})')
    axes[i, 3].set_xlabel('Redshift (z)')

    # Panel 4: f_host_2500 (continuous, decomp_host True)
    d = df_plot[df_plot['decomp_host'] == True]
    vmin, vmax = 0.01, 0.1
    sc0 = axes[i, 4].scatter(
        d['z'], d['apparent_mag_2500'],
        c=np.clip(d['f_host_2500'], vmin, vmax),
        cmap=cmap_host, s=8, alpha=0.5, vmin=vmin, vmax=vmax
    )
    # create colorbar and force its first tick to be vmin (0.01)
    cbar = fig.colorbar(sc0, ax=axes[i, 4])
    #ticks = np.linspace(vmin, vmax, 5)
    #ticks[0] = vmin
    #cbar.set_ticks(ticks)
    #cbar.set_ticklabels([f"{t:.2f}" for t in ticks])
    cbar.set_label(f'f_host_2500 (clipped {vmin}–{vmax})')
    draw_lbol_isocontour(axes[i, 4], df_plot)
    axes[i, 4].set_title(f'Color: f_host_2500 ({title})')
    axes[i, 4].set_xlabel('Redshift (z)')

    # Panel 5: log_lbol (continuous)
    sc_lbol = axes[i, 5].scatter(
        df_plot['z'], df_plot['apparent_mag_2500'],
        c=df_plot['log_lbol'], cmap=cmap_lbol, s=8, alpha=0.5,
        vmin=43, vmax=47
    )
    add_colorbar(axes[i, 5], sc_lbol, 'log_lbol')
    draw_lbol_isocontour(axes[i, 5], df_plot)
    axes[i, 5].set_title(f'Color: log_lbol ({title})')
    axes[i, 5].set_xlabel('Redshift (z)')

    # Panel 6: decomp_host (categorical)
    decomp = df_plot['decomp_host']
    decomp_codes = np.full(len(decomp), 2)  # default NaN
    decomp_codes[np.where(decomp == False)[0]] = 0
    decomp_codes[np.where(decomp == True)[0]]  = 1
    cats = ['False', 'True', 'NaN']
    cols = ['tab:orange', 'tab:blue', 'tab:gray']
    norm, cat_cmap, mappable, ticks, ticklabels = categorical_mappable(cats, cols)
    sc_decomp = axes[i, 6].scatter(
        df_plot['z'], df_plot['apparent_mag_2500'],
        c=decomp_codes, cmap=cat_cmap, norm=norm, s=8, alpha=0.3
    )
    add_colorbar(axes[i, 6], mappable, 'decomp_host', ticks=ticks, ticklabels=ticklabels)
    draw_lbol_isocontour(axes[i, 6], df_plot)
    axes[i, 6].set_title(f'Color: decomp_host ({title})')
    axes[i, 6].set_xlabel('Redshift (z)')

    # Panel 7: npca_qso (categorical; robust to extras/NaN)
    npca = pd.to_numeric(df_plot['npca_qso'], errors='coerce')
    base_vals = [-1, 0, 1, 2, 5, 10]
    present_ints = sorted({int(v) for v in npca.dropna().unique()})
    extras = [v for v in present_ints if v not in base_vals]
    cats_no_nan = base_vals + extras
    nan_code = len(cats_no_nan)
    cats_full = cats_no_nan + ['NaN']
    codes = np.full(npca.shape[0], nan_code, dtype=int)
    mask = npca.notna().to_numpy()
    vals = npca[mask].astype(int).to_numpy()
    map_to_code = {v: i for i, v in enumerate(cats_no_nan)}
    codes[mask] = [map_to_code.get(v, nan_code) for v in vals]
    color_map = {-1: 'tab:gray', 0: 'tab:red', 1: 'tab:orange', 2: 'tab:green', 5: 'tab:blue', 10: 'tab:purple'}
    cols = [color_map.get(v, '0.5') for v in cats_no_nan] + ['0.7']
    norm, cat_cmap, mappable, ticks, ticklabels = categorical_mappable(cats_full, cols)
    sc_npca = axes[i, 7].scatter(
        df_plot['z'], df_plot['apparent_mag_2500'],
        c=codes, cmap=cat_cmap, norm=norm, s=8, alpha=0.3
    )
    add_colorbar(axes[i, 7], mappable, 'npca_qso', ticks=ticks, ticklabels=[str(t) for t in ticklabels])
    draw_lbol_isocontour(axes[i, 7], df_plot)
    axes[i, 7].set_title(f'Color: npca_qso ({title})')
    axes[i, 7].set_xlabel('Redshift (z)')

    # Panel 8: BC (categorical)
    BC = df_plot['BC']
    bc_codes = np.full(len(BC), 2)  # NaN
    bc_codes[np.where(BC == False)[0]] = 0
    bc_codes[np.where(BC == True)[0]]  = 1
    cats = ['False', 'True', 'NaN']
    cols = ['tab:orange', 'tab:blue', 'tab:gray']
    norm, cat_cmap, mappable, ticks, ticklabels = categorical_mappable(cats, cols)
    sc_bc = axes[i, 8].scatter(
        df_plot['z'], df_plot['apparent_mag_2500'],
        c=bc_codes, cmap=cat_cmap, norm=norm, s=8, alpha=0.3
    )
    add_colorbar(axes[i, 8], mappable, 'BC', ticks=ticks, ticklabels=ticklabels)
    draw_lbol_isocontour(axes[i, 8], df_plot)
    axes[i, 8].set_title(f'Color: BC ({title})')
    axes[i, 8].set_xlabel('Redshift (z)')

    # Panel 9: alpha_lambda (continuous)
    alpha = pd.to_numeric(df_plot['alpha_lambda'], errors='coerce')
    vmin_a, vmax_b = np.nanpercentile(alpha, [1, 99])
    sc_alpha = axes[i, 9].scatter(
        df_plot['z'], df_plot['apparent_mag_2500'],
        c=alpha, cmap=cmap_alpha, s=8, alpha=0.5, vmin=vmin_a, vmax=vmax_b
    )
    add_colorbar(axes[i, 9], sc_alpha, 'alpha_lambda')
    draw_lbol_isocontour(axes[i, 9], df_plot)
    axes[i, 9].set_title(f'Color: alpha_lambda ({title})')
    axes[i, 9].set_xlabel('Redshift (z)')

    # Panel 10: sn_median_all (continuous)
    sn_median = pd.to_numeric(df_plot['sn_median_all'], errors='coerce')
    vmin_s, vmax_s = np.nanpercentile(sn_median, [0, 100])
    sc_sn_median = axes[i, 10].scatter(
        df_plot['z'], df_plot['apparent_mag_2500'],
        c=np.clip(sn_median, vmin_s, vmax_s),
        cmap=cmap_sn_median, s=8, alpha=0.3, vmin=vmin_s, vmax=vmax_s
    )
    add_colorbar(axes[i, 10], sc_sn_median, f'sn_median_all (clipped {vmin_s:.2g}–{vmax_s:.2g})')
    draw_lbol_isocontour(axes[i, 10], df_plot)
    axes[i, 10].set_title(f'Color: sn_median_all ({title})')
    axes[i, 10].set_xlabel('Redshift (z)')
    axes[i, 10].set_ylabel('apparent_mag_2500')

    poly_series = df_plot['poly']
    # Normalize to booleans where possible; keep NaN
    poly_codes = np.full(len(poly_series), 2)  # default NaN code
    poly_codes[np.where(poly_series == False)[0]] = 0
    poly_codes[np.where(poly_series == True)[0]]  = 1

    cats = ['False', 'True', 'NaN']
    cols = ['tab:orange', 'tab:blue', 'tab:gray']
    norm, cat_cmap, mappable, ticks, ticklabels = categorical_mappable(cats, cols)

    sc_poly = axes[i, 11].scatter(
        df_plot['z'], df_plot['apparent_mag_2500'],
        c=poly_codes, cmap=cat_cmap, norm=norm, s=8, alpha=0.3
    )
    add_colorbar(axes[i, 11], mappable, 'poly', ticks=ticks, ticklabels=ticklabels)
    draw_lbol_isocontour(axes[i, 11], df_plot)
    axes[i, 11].set_title(f'Color: poly ({title})')
    axes[i, 11].set_xlabel('Redshift (z)')
    axes[i, 11].set_ylabel('apparent_mag_2500')


    # # Panel 12: PL_slope_blue (continuous, zero-centered)
    # plb = pd.to_numeric(df_plot['PL_slope_blue'], errors='coerce')
    # # symmetric clip around 0 using the 99th percentile of |values|
    # amp_b = np.nanpercentile(np.abs(plb.dropna()), 99) if np.isfinite(plb).any() else 1.0
    # vmin_b, vmax_b = -amp_b, amp_b
    # sc_plb = axes[i, 12].scatter(
    #     df_plot['z'], df_plot['apparent_mag_2500'],
    #     c=np.clip(plb, vmin_b, vmax_b),
    #     cmap='viridis', s=8, alpha=0.5, vmin=vmin_b, vmax=vmax_b
    # )
    # add_colorbar(axes[i, 12], sc_plb, f'PL_slope_blue (clipped {vmin_b:.2g}–{vmax_b:.2g})')
    # draw_lbol_isocontour(axes[i, 12], df_plot)
    # axes[i, 12].set_title(f'Color: PL_slope_blue ({title})')
    # axes[i, 12].set_xlabel('Redshift (z)')

    # # Panel 13: PL_slope_red (continuous, zero-centered)
    # plr = pd.to_numeric(df_plot['PL_slope_red'], errors='coerce')
    # amp_r = np.nanpercentile(np.abs(plr.dropna()), 99) if np.isfinite(plr).any() else 1.0
    # vmin_r, vmax_r = -amp_r, amp_r
    # sc_plr = axes[i, 13].scatter(
    #     df_plot['z'], df_plot['apparent_mag_2500'],
    #     c=np.clip(plr, vmin_r, vmax_r),
    #     cmap='viridis', s=8, alpha=0.5, vmin=vmin_r, vmax=vmax_r
    # )
    # add_colorbar(axes[i, 13], sc_plr, f'PL_slope_red (clipped {vmin_r:.2g}–{vmax_r:.2g})')
    # draw_lbol_isocontour(axes[i, 13], df_plot)
    # axes[i, 13].set_title(f'Color: PL_slope_red ({title})')
    # axes[i, 13].set_xlabel('Redshift (z)')

    # # Panel 14: difference PL slopes (continuous, zero-centered)
    # plr = pd.to_numeric(df_plot['alpha_lambda']-df_plot['PL_slope_red'], errors='coerce')
    # amp_r = np.nanpercentile(np.abs(plr.dropna()), 99) if np.isfinite(plr).any() else 1.0
    # vmin_r, vmax_r = -amp_r, amp_r
    # sc_plr = axes[i, 14].scatter(
    #     df_plot['z'], df_plot['apparent_mag_2500'],
    #     c=np.clip(plr, vmin_r, vmax_r),
    #     cmap='coolwarm', s=8, alpha=0.5, vmin=vmin_r, vmax=vmax_r
    # )
    # add_colorbar(axes[i, 14], sc_plr, f'PL_slope_blue - red (clipped {vmin_r:.2g}–{vmax_r:.2g})')
    # draw_lbol_isocontour(axes[i, 14], df_plot)
    # axes[i, 14].set_title(f'Color: PL_slope_blue - red ({title})')
    # axes[i, 14].set_xlabel('Redshift (z)')

    # # Panel 14: PL_break_wave_inbounds (categorical)
    # pl_break_wave_inbounds = df_plot['PL_break_wave_inbounds']
    # pl_break_codes = np.full(len(pl_break_wave_inbounds), 2)  # default NaN code
    # pl_break_codes[np.where(pl_break_wave_inbounds == False)[0]] = 0
    # pl_break_codes[np.where(pl_break_wave_inbounds == True)[0]] = 1

    # cats = ['False', 'True', 'NaN']
    # cols = ['tab:orange', 'tab:blue', 'tab:gray']
    # norm, cat_cmap, mappable, ticks, ticklabels = categorical_mappable(cats, cols)

    # sc_pl_break = axes[i, 15].scatter(
    #     df_plot['z'], df_plot['apparent_mag_2500'],
    #     c=pl_break_codes, cmap=cat_cmap, norm=norm, s=8, alpha=0.3
    # )
    # add_colorbar(axes[i, 15], mappable, 'PL_break_wave_inbounds', ticks=ticks, ticklabels=ticklabels)
    # draw_lbol_isocontour(axes[i, 15], df_plot)
    # axes[i, 15].set_title(f'Color: PL_break_wave_inbounds ({title})')
    axes[i, 15].set_xlabel('Redshift (z)')

    # # Panel 14: PL_break_wave (continuous)
    # pl_break_wave = pd.to_numeric(df_plot['PL_break_wave'], errors='coerce')
    # vmin_bw, vmax_bw = np.nanpercentile(pl_break_wave, [1, 99])
    # sc_pl_break_wave = axes[i, 16].scatter(
    #     df_plot['z'], df_plot['apparent_mag_2500'],
    #     c=pl_break_wave, cmap='viridis', s=8, alpha=0.5, vmin=vmin_bw, vmax=vmax_bw
    # )
    # add_colorbar(axes[i, 16], sc_pl_break_wave, f'PL_break_wave (clipped {vmin_bw:.2g}–{vmax_bw:.2g})')
    # draw_lbol_isocontour(axes[i, 16], df_plot)
    # axes[i, 16].set_title(f'Color: PL_break_wave ({title})')
    # axes[i, 16].set_xlabel('Redshift (z)')


    for j in range(ncols):
        axes[i, j].set_ylim(20, 25)

plt.savefig('plots/m2500_colors.png', dpi=180)
plt.show()
plt.close()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.colors import LogNorm, Normalize

def plot_mag_vs_z_by_npca_bc_poly(
    df,
    *,
    npca_levels=(-1, 0, 1, 2, 5, 10),
    bc_levels=(False, True),
    poly_levels=(False, True),
    cmap_name='viridis',
    point_size=8,
    alpha=0.7,
    robust=True,
    q_lo=0.02,
    q_hi=0.98,
    log_color=False,  # set True if redchi2 spans orders of magnitude
):
    """
    Grid: rows = (BC, poly) combos; cols = npca_qso levels.
    Each panel: apparent_mag_2500 vs z colored by redchi2_conti_full.
    """

    # --- Coercions & finite mask
    df = df.copy()
    # Ensure expected columns exist
    required = {'z', 'apparent_mag_2500', 'redchi2_conti_full', 'npca_qso', 'BC', 'poly'}
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise KeyError(f"Missing required columns: {missing}")

    # Coerce types
    df['npca_qso'] = pd.to_numeric(df['npca_qso'], errors='coerce')
    # BC / poly may come as strings
    for bcol in ['BC', 'poly']:
        if df[bcol].dtype != bool:
            df[bcol] = (df[bcol].astype(str).str.strip().str.lower()
                        .isin(['true', '1', 't', 'yes']))

    # Finite data for plotting
    base_mask = np.isfinite(df['z']) & np.isfinite(df['apparent_mag_2500']) & np.isfinite(df['redchi2_conti_full'])
    data = df.loc[base_mask].copy()

    if data.empty:
        print("No finite data to plot.")
        return

    # Replace non-positive chi2 if we intend to use LogNorm
    if log_color:
        # avoid zeros/negatives for LogNorm
        eps = np.finfo(float).tiny
        data.loc[data['redchi2_conti_full'] <= 0, 'redchi2_conti_full'] = eps

    # Global color scale (robust quantiles or min/max)
    chi = data['redchi2_conti_full'].to_numpy()
    # if robust:
    #     vmin, vmax = np.nanquantile(chi, [q_lo, q_hi])
    #     # Guard against degenerate range
    #     if not np.isfinite(vmin) or not np.isfinite(vmax) or vmin == vmax:
    #         vmin, vmax = np.nanmin(chi), np.nanmax(chi)
    # else:
    #     vmin, vmax = np.nanmin(chi), np.nanmax(chi)
    vmin, vmax = 1, 3

    norm = LogNorm(vmin=max(vmin, np.finfo(float).tiny), vmax=vmax) if log_color else Normalize(vmin=vmin, vmax=vmax)

    # --- Grid layout
    # Row ordering: (BC=False, poly=False), (BC=False, poly=True), (BC=True, poly=False), (BC=True, poly=True)
    row_combos = [(bc, pl) for bc in bc_levels for pl in poly_levels]
    nrows = len(row_combos)
    ncols = len(npca_levels)

    # Reasonable figure size
    fig_w = 4.8 * ncols
    fig_h = 3.6 * nrows
    fig, axes = plt.subplots(nrows, ncols, figsize=(fig_w, fig_h), sharex=True, sharey=True, constrained_layout=True)

    if nrows == 1:
        axes = np.atleast_2d(axes)
    if ncols == 1:
        axes = axes.reshape(nrows, 1)

    cmap = plt.get_cmap(cmap_name)

    # Global axis limits for consistency
    x = data['z'].to_numpy()
    y = data['apparent_mag_2500'].to_numpy()
    # Use robust limits for y (magnitudes often have outliers)
    ylo, yhi = (np.nanquantile(y, [0.01, 0.99]) if robust else (np.nanmin(y), np.nanmax(y)))
    xlo, xhi = (np.nanquantile(x, [0.00, 0.995]) if robust else (np.nanmin(x), np.nanmax(x)))

    for r, (bc, pl) in enumerate(row_combos):
        for c, nq in enumerate(npca_levels):
            ax = axes[r, c]
            sel = (data['npca_qso'] == nq) & (data['BC'] == bc) & (data['poly'] == pl)
            sub = data.loc[sel]

            if sub.empty:
                ax.text(0.5, 0.5, "No data", ha='center', va='center', transform=ax.transAxes, fontsize=10)
                ax.set_xlim(xlo, xhi)
                ax.set_ylim(yhi, ylo)  # mag axis inverted visually
                # Still draw a blank colorbar frame for consistent layout
                sm = plt.cm.ScalarMappable(norm=norm, cmap=cmap)
                sm.set_array([])
                cbar = plt.colorbar(sm, ax=ax)
                cbar.set_label('redchi2_conti_full')
            else:
                sc = ax.scatter(
                    sub['z'].to_numpy(),
                    sub['apparent_mag_2500'].to_numpy(),
                    c=sub['redchi2_conti_full'].to_numpy(),
                    s=point_size,
                    alpha=alpha,
                    cmap=cmap,
                    norm=norm,
                    edgecolors='none',
                )
                ax.set_xlim(xlo, xhi)
                ax.set_ylim(yhi, ylo)  # brighter objects at top (smaller mags)

                # Per-panel colorbar
                cbar = plt.colorbar(sc, ax=ax)
                cbar.set_label('redchi2_conti_full')

            # Titles and panel labels
            label_npca = f"{int(nq)}" if float(nq).is_integer() else str(nq)
            ax.set_title(f"npca={label_npca}  |  BC={bc}  |  poly={pl}", fontsize=12)

            # Axes labels only on left/bottom to reduce clutter
            if c == 0:
                ax.set_ylabel("apparent_mag_2500")
            if r == nrows - 1:
                ax.set_xlabel("z")

            # Light grid
            ax.grid(True, ls=':', lw=0.5, alpha=0.6)
            ax.set_ylim(19, 27)

    #fig.suptitle("apparent_mag_2500 vs redshift, colored by redchi2_conti_full\nRows: (BC, poly); Cols: npca_qso", y=1.02, fontsize=14)
    plt.savefig(f"plots/mag_vs_z_by_npca_bc_poly.png", dpi=150)
    plt.show()
    plt.close()

plot_mag_vs_z_by_npca_bc_poly(df_agn_all, log_color=False)   # or True if chi2 spans orders of magnitude
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def plot_alpha_hist_by_npca_bc_poly(
    df,
    *,
    npca_levels=(-1, 0, 1, 2, 5, 10),
    bc_levels=(False, True),
    poly_levels=(False, True),
    bins=30,
    robust=True,
    q_lo=0.02,
    q_hi=0.98,
    density=False,     # set True to show PDFs instead of counts
    alpha_fill=0.8,    # face alpha for hist bars
    line_width=1.5,    # median/IQR line width
    color='tab:blue',  # bar color
):
    """
    Grid: rows = (BC, poly) combos; cols = npca_qso levels.
    Each panel: histogram of alpha_lambda for the subset.
    """

    df = df.copy()

    # --- Required columns
    required = {'alpha_lambda', 'npca_qso', 'BC', 'poly'}
    missing = [c for c in required if c not in df.columns]
    if missing:
        raise KeyError(f"Missing required columns: {missing}")

    # --- Coerce types
    df['npca_qso'] = pd.to_numeric(df['npca_qso'], errors='coerce')
    for bcol in ['BC', 'poly']:
        if df[bcol].dtype != bool:
            df[bcol] = (df[bcol].astype(str).str.strip().str.lower()
                        .isin(['true', '1', 't', 'yes']))

    # --- Finite data
    m = np.isfinite(df['alpha_lambda']) & np.isfinite(df['npca_qso'])
    data = df.loc[m, ['alpha_lambda', 'npca_qso', 'BC', 'poly']].copy()
    if data.empty:
        print("No finite data to plot.")
        return

    # --- Global bin edges for consistency
    al = data['alpha_lambda'].to_numpy()
    if robust:
        lo, hi = np.nanquantile(al, [q_lo, q_hi])
        if not np.isfinite(lo) or not np.isfinite(hi) or lo == hi:
            lo, hi = np.nanmin(al), np.nanmax(al)
    else:
        lo, hi = np.nanmin(al), np.nanmax(al)
    # pad slightly to avoid edge clipping
    pad = 0.02 * (hi - lo if hi > lo else 1.0)
    lo, hi = lo - pad, hi + pad
    bin_edges = np.linspace(lo, hi, bins + 1)

    # --- Grid layout
    row_combos = [(bc, pl) for bc in bc_levels for pl in poly_levels]
    nrows = len(row_combos)
    ncols = len(npca_levels)

    fig_w = 4.6 * ncols
    fig_h = 3.6 * nrows
    fig, axes = plt.subplots(nrows, ncols, figsize=(fig_w, fig_h),
                             sharex=True, sharey=True, constrained_layout=True)

    if nrows == 1:
        axes = np.atleast_2d(axes)
    if ncols == 1:
        axes = axes.reshape(nrows, 1)

    # --- Plot per cell
    global_ymax = 0.0
    panel_stats = {}  # to normalize y-limits after we know maxima

    for r, (bc, pl) in enumerate(row_combos):
        for c, nq in enumerate(npca_levels):
            ax = axes[r, c]
            mask = (data['BC'] == bc) & (data['poly'] == pl) & (data['npca_qso'] == nq)
            sub = data.loc[mask, 'alpha_lambda']

            label_npca = f"{int(nq)}" if float(nq).is_integer() else str(nq)
            ax.set_title(f"npca={label_npca}  |  BC={bc}  |  poly={pl}", fontsize=11)

            if sub.empty:
                ax.text(0.5, 0.5, "No data", ha='center', va='center',
                        transform=ax.transAxes, fontsize=10)
                panel_stats[(r, c)] = (0, np.nan, np.nan, np.nan, np.nan)
                continue

            counts, _, _ = ax.hist(
                sub.to_numpy(),
                bins=bin_edges,
                density=density,
                histtype='bar',
                edgecolor='white',
                linewidth=0.6,
                color=color,
                alpha=alpha_fill,
            )

            # Stats: N, median, IQR
            N = int(sub.shape[0])
            med = np.nanmedian(sub)
            q25, q75 = np.nanquantile(sub, [0.25, 0.75])

            # Median & IQR markers
            ax.axvline(med, linestyle='-', linewidth=line_width, color='black', alpha=0.9)
            ax.axvline(q25, linestyle='--', linewidth=1.2, color='black', alpha=0.7)
            ax.axvline(q75, linestyle='--', linewidth=1.2, color='black', alpha=0.7)

            # Annotate N and median
            y_top = np.nanmax(counts) if counts.size else 0
            txt = f"N={N}\nmed={med:.3f}\nIQR=[{q25:.3f},{q75:.3f}]"
            ax.text(0.98, 0.95, txt, ha='right', va='top',
                    transform=ax.transAxes, fontsize=9,
                    bbox=dict(boxstyle='round,pad=0.25', fc='white', ec='0.7', alpha=0.8))

            panel_stats[(r, c)] = (y_top, med, q25, q75, N)
            global_ymax = max(global_ymax, y_top)

            # Light grid
            ax.grid(True, ls=':', lw=0.5, alpha=0.6)

            # Axes labels on left/bottom only
            if c == 0:
                ax.set_ylabel("Density" if density else "Count")
            if r == nrows - 1:
                ax.set_xlabel("alpha_lambda")

    # Harmonize y-limits across panels
    if global_ymax == 0:
        global_ymax = 1.0
    for r in range(nrows):
        for c in range(ncols):
            axes[r, c].set_xlim(lo, hi)
            axes[r, c].set_ylim(0, global_ymax * (1.08 if not density else 1.15))

    fig.suptitle("Distribution of α_λ by (BC, poly, npca_qso)\nRows: (BC, poly); Cols: npca_qso",
                 y=1.02, fontsize=14)
    plt.savefig("plots/alpha_lambda_hist_by_npca_bc_poly.png", dpi=150)
    plt.show()
    plt.close()

# Usage
plot_alpha_hist_by_npca_bc_poly(df_agn_all)
```

```python
#!/usr/bin/env python3
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# --------------------- CONFIG ---------------------
CSV_PATH = 'results/data/oct13h_nofhost_oct9b_lagblrband_chisq_plbreakfree_redchi2_newmatch_newredchi2_bc_dustextinction_keepall.csv'
QSOS = [0, 1, 2]
OUT_DIR = "plots/qso_compare_012"
os.makedirs(OUT_DIR, exist_ok=True)

# --------------------- LOAD -----------------------
df = pd.read_csv(CSV_PATH)
df['object_id'] = df['object_id'].astype(str)
# Clip absurdly high χ² values (for stability)
df['redchi2_conti_full'] = df['redchi2_conti_full'].clip(upper=50.0)
# Focus on host-decomposed and requested qso modes
df_h = df[(df['decomp_host'] == True) & (df['npca_qso'].isin(QSOS))].copy()

# Offsets from unity and their absolute values
#   offset = chi2 - 1 (signed)
#   abs_offset = |chi2 - 1| (used for picking the best model)
df_h['offset'] = df_h['redchi2_conti_full'] - 1.0
df_h['abs_offset'] = np.abs(df_h['offset'])

# ------------------- WIDE TABLE -------------------
# One row per object with columns for each qso: offset and |offset|
off_wide = df_h.pivot_table(index='object_id', columns='npca_qso', values='offset', aggfunc='min')
aoff_wide = df_h.pivot_table(index='object_id', columns='npca_qso', values='abs_offset', aggfunc='min')

# Ensure all requested modes exist (fill with NaN if absent), and order them
for q in QSOS:
    if q not in off_wide.columns:
        off_wide[q] = np.nan
    if q not in aoff_wide.columns:
        aoff_wide[q] = np.nan
off_wide = off_wide[QSOS]
aoff_wide = aoff_wide[QSOS]

# Rename columns to readable labels
off_wide = off_wide.rename(columns={0:'offset_q0', 1:'offset_q1', 2:'offset_q2'})
aoff_wide = aoff_wide.rename(columns={0:'abs_offset_q0', 1:'abs_offset_q1', 2:'abs_offset_q2'})

# Combine into a single table for convenience
chi_wide = pd.concat([off_wide, aoff_wide], axis=1)

# ----------------- PAIRWISE Δ-MATRIX -----------------
# Δ_{i→j} = (|offset_i| - |offset_j|) / |offset_j|
# Positive Δ_{i→j} ⇒ model j is preferred over model i (j is closer to unity)
labels = ['q0','q1','q2']
abs_map = {'q0':'abs_offset_q0', 'q1':'abs_offset_q1', 'q2':'abs_offset_q2'}

delta_data = {}
for i in labels:
    for j in labels:
        key = (i, j)
        if i == j:
            delta_data[key] = np.zeros(len(chi_wide), dtype=float)
        else:
            ai = chi_wide[abs_map[i]]
            aj = chi_wide[abs_map[j]]
            with np.errstate(invalid='ignore', divide='ignore'):
                delta_data[key] = (ai - aj) / aj  # stays NaN if aj is NaN

delta_mat = pd.DataFrame(delta_data, index=chi_wide.index)
delta_mat.columns = pd.MultiIndex.from_tuples(delta_mat.columns, names=['from', 'to'])

# --------------- WINNER & MARGIN ----------------
# Winner = argmin over {abs_offset_q0, abs_offset_q1, abs_offset_q2}
vals = chi_wide[['abs_offset_q0','abs_offset_q1','abs_offset_q2']].to_numpy()
vals = np.array(vals, copy=True)
vals[~np.isfinite(vals)] = np.inf
winner_idx = vals.argmin(axis=1)                 # 0→q0, 1→q1, 2→q2
no_data = ~np.isfinite(vals).any(axis=1)
winner_idx[no_data] = -1
map_idx_to_label = {0:'q0', 1:'q1', 2:'q2', -1:np.nan}
winner_label = pd.Series([map_idx_to_label[i] for i in winner_idx], index=chi_wide.index, name='best_qso_label')

# Relative margin vs 2nd best: (χ₂ − χ₁) / χ₂, where χ are abs offsets
sorted_vals = np.sort(vals, axis=1)   # ascending (inf at end)
best = sorted_vals[:, 0]
second = sorted_vals[:, 1]
with np.errstate(invalid='ignore', divide='ignore'):
    best_rel_gain_vs_2nd = (second - best) / second
best_rel_gain_vs_2nd[~np.isfinite(best_rel_gain_vs_2nd)] = np.nan
best_rel_gain_vs_2nd = pd.Series(best_rel_gain_vs_2nd, index=chi_wide.index, name='best_rel_gain_vs_2nd')

# Also expose numeric code for convenience
label_to_num = {'q0':0, 'q1':1, 'q2':2}
winner_qso_num = winner_label.map(label_to_num).rename('best_qso')

# --------------- MERGE BACK TO df_agn (optional) ---------------
# If you already have df_agn in your session, we’ll merge; else we create it from df
df_agn = df.copy() if 'df_agn' not in globals() else df_agn.copy()
df_agn['object_id'] = df_agn['object_id'].astype(str)

# Drop old delta_* columns to avoid collisions
df_agn = df_agn.drop(columns=[c for c in df_agn.columns if c.startswith('delta_')], errors='ignore')
df_agn = df_agn.drop(columns=[c for c in df_agn.columns if c.startswith(('delta_','offset_','abs_offset_','best_'))], errors='ignore')


pairwise = pd.DataFrame({
    'object_id': chi_wide.index,
    # Key pairwise deltas (positive ⇒ RHS preferred):
    'delta_q0_vs_q1': delta_mat[('q0','q1')].values,
    'delta_q0_vs_q2': delta_mat[('q0','q2')].values,
    'delta_q1_vs_q2': delta_mat[('q1','q2')].values,
    # Winners and margins:
    'best_qso': winner_qso_num.values,
    'best_qso_label': winner_label.values,
    'best_rel_gain_vs_2nd': best_rel_gain_vs_2nd.values,
    # Offsets (signed and absolute), useful for diagnostics:
    'offset_q0': chi_wide['offset_q0'].values,
    'offset_q1': chi_wide['offset_q1'].values,
    'offset_q2': chi_wide['offset_q2'].values,
    'abs_offset_q0': chi_wide['abs_offset_q0'].values,
    'abs_offset_q1': chi_wide['abs_offset_q1'].values,
    'abs_offset_q2': chi_wide['abs_offset_q2'].values,
})

df_agn = df_agn.merge(pairwise, on='object_id', how='left')

# ------------------- PLOTTING -------------------
def save_hist(values, xlabel, title, fname, bins=60):
    vals = pd.Series(values).replace([np.inf,-np.inf], np.nan).dropna().values
    if vals.size == 0:
        print(f"[WARN] No finite values for {fname}; skipping.")
        return
    plt.figure(figsize=(7,4))
    plt.hist(vals, bins=bins, alpha=0.85)
    plt.axvline(0.0, linestyle='--')
    plt.xlabel(xlabel)
    plt.ylabel('Count')
    plt.title(title)
    plt.tight_layout()
    plt.savefig(os.path.join(OUT_DIR, fname), dpi=150)
    plt.close()

def save_scatter(x, y, xl, yl, title, fname):
    x = pd.Series(x).replace([np.inf,-np.inf], np.nan).to_numpy()
    y = pd.Series(y).replace([np.inf,-np.inf], np.nan).to_numpy()
    m = np.isfinite(x) & np.isfinite(y)
    if not np.any(m):
        print(f"[WARN] No paired finite values for {fname}; skipping.")
        return
    x = x[m]; y = y[m]
    lim = np.nanmax([np.nanpercentile(np.abs(x), 99.5),
                     np.nanpercentile(np.abs(y), 99.5)])
    lim = max(lim, 1e-3)
    plt.figure(figsize=(5.6,5.6))
    plt.axline((0,0), slope=1.0, linestyle='--')   # 1:1 line
    plt.scatter(x, y, s=8, alpha=0.65)
    plt.xlim(-lim, lim); plt.ylim(-lim, lim)       # symmetric for signed offsets
    plt.xlabel(xl); plt.ylabel(yl)
    plt.title(title)
    plt.tight_layout()
    plt.savefig(os.path.join(OUT_DIR, fname), dpi=150)
    plt.close()

# 1) Histograms of Δ (positive ⇒ RHS preferred)
save_hist(df_agn['delta_q0_vs_q1'], r'$\Delta_{0\rightarrow1}$',
          'Histogram of Δ (q0→q1)', 'hist_delta_q0_to_q1.png')
save_hist(df_agn['delta_q0_vs_q2'], r'$\Delta_{0\rightarrow2}$',
          'Histogram of Δ (q0→q2)', 'hist_delta_q0_to_q2.png')
save_hist(df_agn['delta_q1_vs_q2'], r'$\Delta_{1\rightarrow2}$',
          'Histogram of Δ (q1→q2)', 'hist_delta_q1_to_q2.png')

# 2) Scatter of signed offsets (offset_i vs offset_j)
save_scatter(chi_wide['offset_q0'], chi_wide['offset_q1'],
             '|χ²−1| sign kept (q0)', '|χ²−1| sign kept (q1)',
             'Offset (χ²−1): q0 vs q1', 'scatter_offset_q0_vs_q1.png')
save_scatter(chi_wide['offset_q0'], chi_wide['offset_q2'],
             '|χ²−1| sign kept (q0)', '|χ²−1| sign kept (q2)',
             'Offset (χ²−1): q0 vs q2', 'scatter_offset_q0_vs_q2.png')
save_scatter(chi_wide['offset_q1'], chi_wide['offset_q2'],
             '|χ²−1| sign kept (q1)', '|χ²−1| sign kept (q2)',
             'Offset (χ²−1): q1 vs q2', 'scatter_offset_q1_vs_q2.png')

# 3) Bar chart of winners
if 'best_qso' in df_agn.columns:
    counts = df_agn['best_qso'].value_counts().reindex([0,1,2]).fillna(0).astype(int)
    plt.figure(figsize=(6,4))
    plt.bar(['q0','q1','q2'], counts.values)
    plt.xlabel('Best (lowest |χ²−1|)')
    plt.ylabel('Number of objects')
    plt.title('Winning model counts (decomp_host=True)')
    plt.tight_layout()
    plt.savefig(os.path.join(OUT_DIR, "bar_best_counts.png"), dpi=150)
    plt.close()

# 4) Histogram of winner margins
save_hist(best_rel_gain_vs_2nd, 'Relative gain of winner vs second ((χ₂−χ₁)/χ₂)',
          'Winner margin', 'hist_winner_rel_gain.png')

# 5) Heatmap of median Δ across objects
M = np.zeros((3,3), dtype=float)
for i, a in enumerate(labels):
    for j, b in enumerate(labels):
        if a == b:
            M[i, j] = 0.0
        else:
            col = (a, b)
            vals = delta_mat[col].replace([np.inf, -np.inf], np.nan).dropna().values
            M[i, j] = np.median(vals) if vals.size else np.nan

plt.figure(figsize=(5.2,4.8))
im = plt.imshow(M, origin='upper')
plt.xticks(range(3), labels)
plt.yticks(range(3), labels)
plt.colorbar(im, label='Median Δ_{i→j}\n(positive ⇒ j preferred)')
plt.title('Median pairwise Δ (decomp_host=True)')
for i in range(3):
    for j in range(3):
        txt = f"{M[i,j]:.2f}" if np.isfinite(M[i,j]) else "NaN"
        plt.text(j, i, txt, ha='center', va='center')
plt.tight_layout()
plt.savefig(os.path.join(OUT_DIR, "heatmap_median_delta.png"), dpi=150)
plt.close()

print(f"[OK] Comparison complete. Outputs in: {OUT_DIR}")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

band_colors = {
    'u': 'tab:blue',
    'g': 'tab:green', 
    'r': 'tab:orange', 
    'i': 'tab:red', 
    'z': 'tab:brown', 
    'y': 'tab:gray'
}

df_residuals = pd.read_csv(
    'plots/hubble/oct5a_preview_fhost0.01_oct3g_zspecoct5b_redint_plslope0_redchicut_dev_zin/Flatw0waCDM_joint_dev/residuals.csv'
)

def bands_bluer_than_lyman_alpha(z, buffer=100):
    lyman_alpha = 1216.0
    bands = {
        'u': {'lambda_eff': 3551.0},
        'g': {'lambda_eff': 4686.0},
        'r': {'lambda_eff': 6165.0},
        'i': {'lambda_eff': 7481.0},
        'z': {'lambda_eff': 8931.0},
        'y': {'lambda_eff': 9700.0},
    }
    bluer_bands = []
    for band, props in bands.items():
        rest_lambda_eff = props['lambda_eff'] / (1.0 + z)
        if rest_lambda_eff < (lyman_alpha - buffer):
            bluer_bands.append(band)
    return bluer_bands
def sdss_bands_affected_by_lya(z, buffer=100.0):
    """
    SDSS ugriz bands whose *rest-frame blue edge* falls below 1216+buffer Å,
    i.e., likely contaminated by Lyα line/forest.

    Args:
        z (float): source redshift
        buffer (float): safety margin in Å (typ. 50–200)

    Returns:
        list[str]: bands to drop (contaminated)
    """
    # SDSS (full transmission) edges in observed Å (λ_min, λ_max)
    edges_obs = {
        "u": (3055.11, 4030.64),
        "g": (3797.64, 5553.04),
        "r": (5418.23, 6994.42),
        "i": (6692.41, 8400.32),
        "z": (7964.70, 10873.33),
    }

    cutoff = 1216.0 + buffer
    affected = []
    for b, (lo_obs, _hi_obs) in edges_obs.items():
        lo_rest = lo_obs / (1.0 + z)
        if lo_rest < cutoff:
            affected.append(b)
    return affected


def sdss_bands_safe_from_lya(z, buffer=100.0):
    """Complement: bands whose rest-frame blue edge is ≥ 1216+buffer Å."""
    cutoff = 1216.0 + buffer
    edges_obs = {
        "u": (3055.11, 4030.64),
        "g": (3797.64, 5553.04),
        "r": (5418.23, 6994.42),
        "i": (6692.41, 8400.32),
        "z": (7964.70, 10873.33),
    }
    return [b for b, (lo_obs, _) in edges_obs.items() if (lo_obs / (1.0 + z)) >= cutoff]

# ---------- binning (unchanged) ----------
bins = np.arange(df_residuals['z'].min(), df_residuals['z'].max() + 0.2, 0.2)
bin_centers = 0.5 * (bins[:-1] + bins[1:])
residuals_binned, residuals_std, counts = [], [], []
for i in range(len(bins) - 1):
    mask = (df_residuals['z'] >= bins[i]) & (df_residuals['z'] < bins[i+1])
    bin_data = df_residuals.loc[mask, 'residuals']
    if len(bin_data) > 0:
        residuals_binned.append(bin_data.mean())
        residuals_std.append(bin_data.std())
        counts.append(len(bin_data))
    else:
        residuals_binned.append(np.nan)
        residuals_std.append(np.nan)
        counts.append(0)
yerr = np.array(residuals_std) / np.sqrt(np.maximum(np.array(counts), 1))

# ---------- plotting ----------
fig, ax = plt.subplots(figsize=(7,5))
ax.scatter(df_residuals['z'], df_residuals['residuals'], s=10, alpha=0.6,
           edgecolor='k', color='k', linewidth=0.3, label='All')
ax.errorbar(bin_centers, residuals_binned, yerr=yerr, fmt='o', color='red',
            label='Binned (Δz=0.2)', capsize=3)

ax.set_xlabel('Redshift (z)')
ax.set_ylabel('Residuals')
ax.set_title('Residuals vs Redshift (binned in Δz=0.2)')
ax.set_ylim(-0.5, 0.5)

# ---------- shade dropped-band regions using the provided function ----------
bands_order = ['u', 'g', 'r', 'i', 'z']  # SDSS
#band_colors = {'u':'#7e57c2', 'g':'#66bb6a', 'r':'#ef5350', 'i':'#8d6e63', 'z':'#616161'}
zmin = float(np.nanmin(df_residuals['z']))
zmax = float(np.nanmax(df_residuals['z']))
zgrid = np.linspace(zmin, zmax, 2000)

def first_drop_z(band, buffer=100):
    dropped = np.array([band in sdss_bands_affected_by_lya(z, buffer=buffer) for z in zgrid])
    if not dropped.any():
        return None
    # First z where it becomes dropped; monotonic thereafter
    return zgrid[np.argmax(dropped)]

# draw spans and tiny vertical labels
legend_patches = []
for b in bands_order:
    z_start = first_drop_z(b, buffer=100)
    print(f"Band {b} drops at z >= {z_start}")
    if z_start is not None and z_start < zmax:
        ax.axvspan(z_start, zmax, alpha=0.40, color=band_colors[b], lw=0)
        # tiny vertical label near top
        ypos = ax.get_ylim()[1] * 0.97
        ax.text(z_start + 0.01*(zmax - zmin), ypos, f'{b}-drop',
                rotation=90, va='top', ha='left', fontsize=6, color=band_colors[b])
        legend_patches.append(Patch(facecolor=band_colors[b], alpha=0.25, label=f'{b} dropped'))

# merge legends
handles, labels = ax.get_legend_handles_labels()
handles += legend_patches
labels  += [p.get_label() for p in legend_patches]
ax.legend(handles, labels, loc='best', fontsize=10, frameon=True)

plt.tight_layout()
plt.show()
```

```python
import numpy as np
from scipy.stats import gaussian_kde, chi2
from scipy.ndimage import gaussian_filter
import matplotlib.pyplot as plt

plt.figure(figsize=(6,5))

# --- scatter: light blue points ---
x = df_agn['log_sigma_UV'].to_numpy()
y = df_agn['log_tau_UV_RF'].to_numpy()
mask = np.isfinite(x) & np.isfinite(y)
x, y = x[mask], y[mask]

plt.scatter(
    x, y,
    c='k',
    alpha=0.7,
    s=14,
    edgecolors='none'
)

# --- KDE grid with generous padding to avoid open (clipped) outer contours ---
# (Use very wide quantiles + extra pad to keep 3σ closed.)
q_lo, q_hi = 0.001, 0.999
xq0, xq1 = np.quantile(x, [q_lo, q_hi])
yq0, yq1 = np.quantile(y, [q_lo, q_hi])
pad_x = 0.25 * (xq1 - xq0)
pad_y = 0.25 * (yq1 - yq0)
xmin, xmax = xq0 - pad_x, xq1 + pad_x
ymin, ymax = yq0 - pad_y, yq1 + pad_y

xx, yy = np.mgrid[xmin:xmax:256j, ymin:ymax:256j]
kde = gaussian_kde(np.vstack([x, y]))
zz = kde(np.vstack([xx.ravel(), yy.ravel()])).reshape(xx.shape)

# Small visual smoothing helps remove jagged edges without changing the mass ranking
zz = gaussian_filter(zz, sigma=0.8)

# --- Highest-density (HPD) contour levels for 1, 2, 3 "sigma" in 2D ---
# In 2D, mass inside radius r is chi2.cdf(r^2, df=2).
sigma = np.array([1.0, 2.0, 3.0])
probs = chi2.cdf(sigma**2, df=2)  # ≈ [0.3935, 0.8647, 0.9889]

z_sort = np.sort(zz.ravel())[::-1]
cdf = np.cumsum(z_sort) / z_sort.sum()
thr = np.interp(probs, cdf, z_sort)

# Matplotlib expects increasing levels (low->high density) i.e., [3σ, 2σ, 1σ]
levels = np.sort(thr)
# Ensure strict monotonicity (avoids "Non-increasing" due to ties)
levels = np.maximum.accumulate(np.nextafter(levels, np.inf))

# Style mapping for ascending levels: [3σ, 2σ, 1σ]
colors = ['m', 'm', 'm']      # 3σ black, 2σ light gray, 1σ dark gray
linestyles = ['--', '--', '--']             # 3σ dashed, 2σ solid, 1σ solid
linewidths = [2, 2, 2.0]

plt.contour(xx, yy, zz, levels=levels, alpha=0.9,
            colors=colors, linestyles=linestyles, linewidths=linewidths)

plt.xlabel(r'$\log_{10}\,\sigma_{\rm UV}$')
plt.ylabel(r'$\log_{10}\,\tau_{\rm UV,RF}$')
#plt.title(r'$\log_{10}\,\tau_{\rm UV,RF}$ vs $\log_{10}\,\sigma_{\rm UV}$')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse
from scipy.stats import chi2

# --- Data matrix (drop NaNs) ---
X = df_agn[['log_sigma_UV','log_tau_UV_RF']].dropna().values
x = X[:,0]; y = X[:,1]

# --- Mahalanobis ellipse params ---
alpha = 0.99  # confidence level for the kept "bulk" (change to 0.95 if you want tighter)
mean = X.mean(axis=0)
cov  = np.cov(X, rowvar=False)
cov_inv = np.linalg.inv(cov)
chi2_thr = chi2.ppf(alpha, df=2)  # threshold on squared Mahalanobis distance

# distances
dX = X - mean
d2 = np.einsum('ij,jk,ik->i', dX, cov_inv, dX)  # squared Mahalanobis distance
inside = d2 <= chi2_thr
outside = ~inside

# --- Ellipse geometry (principal axes of covariance scaled to chosen level) ---
eigvals, eigvecs = np.linalg.eigh(cov)        # ascending order
order = eigvals.argsort()[::-1]
eigvals = eigvals[order]
eigvecs = eigvecs[:,order]

scale = np.sqrt(chi2_thr)
width, height = 2*scale*np.sqrt(eigvals)      # full axis lengths
angle = np.degrees(np.arctan2(eigvecs[1,0], eigvecs[0,0]))

print(f"width={width:.3f}, height={height:.3f}, angle={angle:.1f} deg")

# --- Plot ---
plt.figure(figsize=(7,6))

plt.scatter(x[inside], y[inside],
            c='gray', alpha=0.6, edgecolor='k', s=20, label='Inside ellipse')

plt.scatter(x[outside], y[outside],
            c='red', alpha=0.9, edgecolor='k', s=25, label='Outside ellipse')

ell = Ellipse(xy=mean, width=width, height=height, angle=angle,
              fill=False, lw=2.0, ls='--', edgecolor='black', zorder=3)
plt.gca().add_patch(ell)

plt.xlabel('log_sigma_UV')
plt.ylabel('log_tau_UV_RF')
plt.title('log_tau_UV_RF vs log_sigma_UV')

# clean legend
plt.legend(title=f'{int(alpha*100)}% Mahalanobis ellipse', frameon=True)
plt.tight_layout()
plt.show()
```

```python
def plot_predicted_L2500_vs_sigmahat(
    flat_samples, df_agn, cosmo_model, z_pivot_agn,
    plot_path='plots/hubble', show=False, debias=True, dms=None,
    show_residuals=False
):
    d = df_agn.copy()

    # --- Thinning for speed ---
    n_samples = int(flat_samples.shape[0])
    thin_factor = max(1, n_samples // 500)
    flat_samples = flat_samples[::thin_factor]

    # --- Indices & parameter names ---
    priors, model_labels, model_labels_latex = get_model_params(cosmo_model)
    param_indices = {name: model_labels.index(name) for name in model_labels}

    # --- Pack obs/errs/pivots once ---
    agn_obs_arr, agn_err_arr, agn_pivot_arr = agn_model_pack_obs(d)

    def compute_xlog_and_err(params_arr, obs_arr, err_arr, pivots_array):
        M0_agn = params_arr[agn_model_pidx["M0_agn"]]
        predicted_M = M_model_agn(params_arr, obs_arr, pivots_array) - M0_agn
        x_log = predicted_M

        predited_M_err = M_model_agn_err(params_arr, obs_arr, err_arr, pivots_array)
        x_log_err = predited_M_err

        return x_log, x_log_err

    # --- Posterior medians (for cosmology & placing data in x) ---
    results = {key: np.median(flat_samples[:, i]) for i, key in enumerate(model_labels)}
    if cosmo_model == 'FlatwCDM':
        cosmo = FlatwCDM(H0=results['H0'], Om0=results['Om0'], w0=results['w0'])
    elif cosmo_model == 'FlatwpwaCDM':
        cosmo = FlatwpwaCDM(H0=results['H0'], Om0=results['Om0'], wp=results['wp'], wa=results['wa'], zp=z_pivot_agn)
    elif cosmo_model == 'Flatw0waCDM':
        cosmo = Flatw0waCDM(H0=results['H0'], Om0=results['Om0'], w0=results['w0'], wa=results['wa'])
    elif cosmo_model == 'FlatLambdaCDM':
        cosmo = FlatLambdaCDM(H0=results['H0'], Om0=results['Om0'])
    else:
        raise ValueError(f"Unknown cosmological model: {cosmo_model}")

    # --- y-data: log10 L_2500 (with optional debias) ---
    if debias:
        dm_interp = make_dm_function(d["apparent_mag_2500"].values, d['z'].values, dms)
        pts = np.column_stack([d['z'], d['apparent_mag_2500']])
        actual_M2500 = (d['apparent_mag_2500'] - dm_interp(pts)) - cosmo.distmod(d['z']).value
    else:
        actual_M2500 = d['apparent_mag_2500'] - cosmo.distmod(d['z']).value
    actual_logL2500 = -0.4 * (actual_M2500 - 90.0)

    # y measurement uncertainty from apparent magnitudes (propagated into log10 L)
    y_log_meas_err = 0.4 * np.asarray(d['apparent_mag_2500_err'].fillna(0.0))

    # --- x for data (full predictor) ---
    median_params_arr = agn_model_pack_params(results)
    x_log_data, x_log_err_data = compute_xlog_and_err(median_params_arr, agn_obs_arr, agn_err_arr, agn_pivot_arr)
    x_data = 10**x_log_data
    x_lower = 10**(x_log_data - x_log_err_data)
    x_upper = 10**(x_log_data + x_log_err_data)
    xerr_asym = np.vstack((x_data - np.maximum(x_lower, 1e-300),
                           np.maximum(x_upper, x_data) - x_data))

    # --- Grid for model curves/band ---
    x_lo = np.percentile(x_log_data - x_log_err_data, 0) - 0.05
    x_hi = np.percentile(x_log_data + x_log_err_data, 100) + 0.05
    x_log_grid = np.linspace(x_lo, x_hi, 200)
    x_grid = 10**x_log_grid

    # --- Predict logL on grid for each posterior sample (we'll only use quantiles + median line) ---
    ylog_grid_by_sample = []
    for s in flat_samples:
        sample_params = {k: s[param_indices[k]] for k in model_labels}
        aparr = agn_model_pack_params(sample_params)
        M0 = aparr[agn_model_pidx["M0_agn"]]
        b = -0.4 * (M0 - 90.0)
        ylog_grid_by_sample.append(b + (-0.4) * x_log_grid)

    ylog_grid_by_sample = np.asarray(ylog_grid_by_sample)  # (nsamp, ngrid)
    ylog_med  = np.median(ylog_grid_by_sample, axis=0)
    ylog_low  = np.percentile(ylog_grid_by_sample, 16, axis=0)
    ylog_high = np.percentile(ylog_grid_by_sample, 84, axis=0)
    y_hi = np.max(ylog_high) + 0.05
    y_lo = np.min(ylog_low)  - 0.05

    # --- Residuals vs median model (log space) ---
    f_med  = interp1d(x_log_grid, ylog_med,  bounds_error=False, fill_value='extrapolate')
    f_low  = interp1d(x_log_grid, ylog_low,  bounds_error=False, fill_value='extrapolate')
    f_high = interp1d(x_log_grid, ylog_high, bounds_error=False, fill_value='extrapolate')

    model_logL_at_data = f_med(x_log_data)
    residuals = actual_logL2500 - model_logL_at_data

    # Error budget in log space for residuals: model spread + x-prop + mag error
    obs_err = 0.5 * (f_high(x_log_data) - f_low(x_log_data))
    propagated_err = 0.4 * np.abs(x_log_err_data)
    total_err = np.sqrt(obs_err**2 + propagated_err**2 + y_log_meas_err**2)

    # --- Figure scaffold ---
    color = 'm'
    if show_residuals:
        fig = plt.figure(figsize=(8, 8))
        gs = gridspec.GridSpec(2, 1, height_ratios=[3, 1], hspace=0.05)
        ax = fig.add_subplot(gs[0])
        ax_res = fig.add_subplot(gs[1], sharex=ax)
    else:
        fig, ax = plt.subplots(figsize=(8, 6))
        ax_res = None

    # --- Data with x & y error bars (y errors converted to linear space) ---
    key = 'alpha_lambda'
    color_key = d[key] if key in d.columns else np.zeros(len(d))
    yerr_linear = 10**actual_logL2500 * np.log(10) * y_log_meas_err

    # Solid vs open AGN markers by z (both main and inset)
    mask_in  = d["z"].between(0.44, 3.16)
    mask_out = ~mask_in

    # AGN (inside)
    ax.errorbar(
        x_data[mask_in], 10**actual_logL2500[mask_in], xerr=xerr_asym[:, mask_in], yerr=yerr_linear[mask_in],
        fmt='o', linestyle='none', markersize=3,
        mfc="black", mec="none",
        ecolor="#666666", elinewidth=1.1,
        alpha=0.3, zorder=0, label="AGN"
    )
    # AGN (outside, open)
    ax.errorbar(
        x_data[mask_out], 10**actual_logL2500[mask_out], xerr=xerr_asym[:, mask_out], yerr=yerr_linear[mask_out],
        fmt='o', linestyle='none', markersize=3, mfc='none', mec="k", alpha=0.3,
        ecolor="#666666", elinewidth=1.1, zorder=0
    )

    # ax.scatter(
    #     x_data, 10**actual_logL2500,
    #     s=12, c="black", marker="o", edgecolors="none", alpha=0.8, zorder=1, label="AGN"
    # )
    # ax.errorbar(
    #     x_data, 10**actual_logL2500, xerr=xerr_asym, yerr=yerr_linear,
    #     fmt='none', linestyle='none',
    #     ecolor="#666666", elinewidth=1.1,
    #     alpha=0.3, zorder=0
    # )
    # --- Model: one line (median) + 68% band ---
    ax.fill_between(x_grid, 10**ylog_low, 10**ylog_high, color=color, alpha=0.18, zorder=9,)
    ax.plot(x_grid, 10**ylog_med, color=color, lw=2.0, zorder=10, label='best-fit model')

    # --- Axes & labels ---
    ax.set_ylabel(r'$L_{2500}$ (erg s$^{-1})$')
    ax.set_xscale('log')
    ax.set_yscale('log')
    # ax.set_xlim((3e-7, 7e5))
    # ax.set_ylim((7e41, 2e47))
    ax.set_xlim((10**x_lo, 10**x_hi))
    ax.set_ylim((10**y_lo, 10**y_hi))
    
    ax.xaxis.set_major_locator(LogLocator(base=10.0))
    ax.xaxis.set_minor_locator(LogLocator(base=10.0, subs=np.arange(2, 10) * 0.1, numticks=100))


    ax.legend()

    if show_residuals and ax_res is not None:
        sc = ax_res.scatter(x_data, residuals, s=10, alpha=0.7, c=color_key, cmap='bwr',
                            edgecolor='k', lw=0.5, zorder=5)
        cbar = plt.colorbar(sc, ax=ax_res, orientation='vertical'); cbar.set_label(key)
        ax_res.errorbar(x_data, residuals, yerr=total_err,
                        fmt='none', alpha=0.25, lw=1.2, capsize=2.5, capthick=1, color='k', zorder=4)
        ax_res.axhline(0, color=color, linestyle='--', zorder=3)
        ax_res.set_ylabel('Residuals (log)')
        ax_res.set_xlabel(r'$x=\sigma^\alpha\,\tau^\beta\,10^{\gamma\,\mathrm{logistic}(\alpha_\nu)}$')
        ax_res.set_xscale('log')
        ax_res.set_ylim(-2.2, 2.2)
        plt.setp(ax.get_xticklabels(), visible=False)
    else:
        ax.set_xlabel(r'$(\sigma/\sigma_{\mathrm{p}})^{\alpha}(\tau/\tau_{\mathrm{p}})^{\beta}$')
        plt.setp(ax.get_xticklabels(), visible=True)


    os.makedirs(plot_path, exist_ok=True)
    os.makedirs(os.path.join(plot_path, "pdf"), exist_ok=True)
    if debias:
        plt.savefig(os.path.join(plot_path, "predicted_L2500_vs_fullcorr_band_debiased.png"), dpi=300)
        plt.savefig(os.path.join(plot_path, "pdf", "predicted_L2500_vs_fullcorr_band_debiased.pdf"), dpi=600)
    else:
        plt.savefig(os.path.join(plot_path, "predicted_L2500_vs_fullcorr_band.png"), dpi=300)
        plt.savefig(os.path.join(plot_path, "pdf", "predicted_L2500_vs_fullcorr_band.pdf"), dpi=600)

    if show:
        plt.show()
    plt.close()

plot_predicted_L2500_vs_sigmahat(
    flat_samples_w0wa, df_agn, cosmo_model_w0wa, z_pivot_agn,
    plot_path='plots/hubble', show=True, debias=True, dms=dmag_corr_w0wa,
    show_residuals=False
)
```

```python
import math
from typing import Tuple, Optional

def _round_sig(x: float, sig: int) -> float:
    if x == 0 or not math.isfinite(x):
        return x
    return round(x, -int(math.floor(math.log10(abs(x)))) + (sig - 1))

def format_value_uncertainty(
    median: float,
    err: Optional[float] = None,
    *,
    twosig_when=(1, 2, 3),
    sci: bool = True,
    latex: bool = True,
    unit: Optional[str] = None,
    max_decimals_no_sci: int = 2,          # <- cap decimals in non-sci mode
    prefer_fewer_decimals: bool = True     # <- use 1 sig fig if 2 would exceed the cap
) -> Tuple[float, Optional[float], int, str]:
    """
    Returns (value_rounded, err_rounded, exponent, formatted_string).
    - Uncertainty: 1 sig fig, except 2 when leading digit in twosig_when,
      but if that would exceed `max_decimals_no_sci` in non-sci mode, fall back to 1.
    - Value rounded to same decimal place as the (rounded) uncertainty.
    - No '±' when err is None.
    - Avoid sci for 0.1 <= |value| < 10 (plain decimals).
    - Special-case: fold 1×10^{-1} or 1×10^{-2} into 0.1 / 0.01.
    """
    # choose exponent (initial)
    if median == 0 or not math.isfinite(median):
        exponent = 0
    else:
        exponent = int(math.floor(math.log10(abs(median)))) if sci else 0

    # avoid sci in [0.1, 10)
    if sci and 0.1 <= abs(median) < 10:
        exponent = 0

    scale = 10.0 ** exponent if exponent != 0 else 1.0
    v = median / scale

    def decimals_needed(x: float) -> int:
        if x == 0: return 0
        return max(0, -int(math.floor(math.log10(abs(x)))))  # e.g. 0.02 -> 2

    if err is None:
        v_rounded = _round_sig(v, 3) if exponent != 0 else _round_sig(v, 3)
        e_rounded = None
    else:
        if err < 0 or not math.isfinite(err):
            raise ValueError("err must be finite and non-negative.")
        e = err / scale

        if e == 0:
            e_rounded = 0.0
            v_rounded = round(v, 6)
        else:
            # candidate with 2 sig figs?
            e1 = _round_sig(e, 1)
            lead = int(abs(e1) / (10 ** math.floor(math.log10(abs(e1)))))
            use_two = lead in twosig_when

            # if non-sci and two sig figs would exceed decimal cap, fall back to 1
            if use_two:
                e_two = _round_sig(e, 2)
                if exponent == 0 and prefer_fewer_decimals and decimals_needed(e_two) > max_decimals_no_sci:
                    use_two = False

            sig_unc = 2 if use_two else 1
            e_rounded = _round_sig(e, sig_unc)

            # round value to same decimal place as e_rounded
            digits = decimals_needed(e_rounded)
            v_rounded = round(v, digits)

            # re-enforce after potential order change
            e_rounded = _round_sig(e, sig_unc)
            digits = decimals_needed(e_rounded)
            v_rounded = round(v, digits)

    # special-case: (±)1×10^{-1 or -2} -> fold to decimal
    if sci and exponent in (-1, -2) and (abs(v_rounded) == 1.0):
        v_rounded *= 10 ** exponent
        if err is not None:
            e_rounded *= 10 ** exponent
        exponent = 0
        scale = 1.0

    # build string
    if latex:
        if exponent != 0:
            s = (f"{v_rounded}\\times 10^{{{exponent}}}"
                 if err is None else f"({v_rounded} \\pm {e_rounded})\\times 10^{{{exponent}}}")
        else:
            s = (f"{v_rounded}" if err is None else f"{v_rounded} \\pm {e_rounded}")
        if unit:
            s += f"\\,\\mathrm{{{unit}}}"
    else:
        if exponent != 0:
            s = (f"{v_rounded}×10^{exponent}"
                 if err is None else f"({v_rounded} ± {e_rounded})×10^{exponent}")
        else:
            s = (f"{v_rounded}" if err is None else f"{v_rounded} ± {e_rounded}")
        if unit:
            s += f" {unit}"

    val_out = v_rounded * (10 ** exponent if exponent != 0 else 1.0)
    err_out = (e_rounded * (10 ** exponent) if (err is not None and exponent != 0) else (e_rounded if err is not None else None))
    
    return s
    return val_out, err_out, exponent, s



def write_results_tex_variables(df_agn, flat_samples, cosmo_model, compare_r, z_pivot_agn, write_path):
    """
    Write key cosmological parameters to a LaTeX file as \newcommand definitions.

    Parameters
    ----------
    flat_samples : (N, P) array
        Flattened MCMC samples: N total draws by P parameters.
    cosmo_model : str
        Cosmological model string.
    z_pivot_agn : float
        Pivot redshift for w0/wa conversion if applicable.
    write_path : str
        Directory path to write the LaTeX file (filename is 'cosmo_params.tex').
    """

    obs_arr, err_arr, pivots_arr = agn_model_pack_obs(df_agn)

    log_sigma_UV_pivot  = pivots_arr[agn_model_oidx["log_sigma_UV"]]
    log_tau_UV_RF_pivot = pivots_arr[agn_model_oidx["log_tau_UV_RF"]]


    priors, model_labels, _ = get_model_params(cosmo_model)
    flat_samples = np.asarray(flat_samples)

    results = {key: sym_percentile(flat_samples[:, i])
               for i, key in enumerate(model_labels)}

    # Write to LaTeX file
    lines = []
    lines.append(r"% Auto-generated cosmological parameters from MCMC samples")
    lines.append(r"% Do not edit by hand; regenerate with write_results_tex_variables()")
    lines.append(r"\newcommand{\resultNumAGN}{%d}" % len(df_agn))
    lines.append(r"\newcommand{\resultSigma}{\ensuremath{%s}}" % format_value_uncertainty(compare_r["sigma_from_odds_two_sided"], None))

    lines.append(r"\newcommand{\resultAlphaAGN}{\ensuremath{%s}}" % format_value_uncertainty(results['alpha_agn'][0], results['alpha_agn'][1]))
    lines.append(r"\newcommand{\resultBetaAGN}{\ensuremath{%s}}" % format_value_uncertainty(results['beta_agn'][0], results['beta_agn'][1]))

    lines.append(r"\newcommand{\resultSigmaUVPivot}{\ensuremath{%s}}" % format_value_uncertainty(10**log_sigma_UV_pivot, None))
    lines.append(r"\newcommand{\resultTauUVRFPivot}{\ensuremath{%s}}" % format_value_uncertainty(10**log_tau_UV_RF_pivot, None))



    M0_agn_samples = flat_samples[:, model_labels.index('M0_agn')]
    alpha_agn_samples = flat_samples[:, model_labels.index('alpha_agn')]
    beta_agn_samples = flat_samples[:, model_labels.index('beta_agn')]

    alpha_AGN_L_samples = alpha_agn_samples * (-1/2.5)
    beta_AGN_L_samples  = beta_agn_samples * (-1/2.5)
    L_intercept_samples = np.power(10, (90-M0_agn_samples)/2.5)

    L_intercept, L_intercept_err = sym_percentile(L_intercept_samples)
    val, err = L_intercept, L_intercept_err

    # --- Extract exponent from the value ---
    exp = int(np.floor(np.log10(abs(val)))) if val != 0 else 0
    scale = 10.0**exp

    mant_val = val / scale
    mant_err = err / scale

    lines.append(
        r"\newcommand{\resultLIntercept}{\ensuremath{%s}}" %
        format_value_uncertainty(*sym_percentile(L_intercept_samples), unit=r"erg\,s^{-1}")
    )

    lines.append(r"\newcommand{\resultAlphaAGNL}{\ensuremath{%s}}" % format_value_uncertainty(*sym_percentile(alpha_AGN_L_samples)))
    lines.append(r"\newcommand{\resultBetaAGNL}{\ensuremath{%s}}" % format_value_uncertainty(*sym_percentile(beta_AGN_L_samples)))

    hd_scatter_samples = np.exp(flat_samples[:, model_labels.index('log_f')])
    lines.append(r"\newcommand{\resultScatterHD}{\ensuremath{%s}}" % format_value_uncertainty(*sym_percentile(hd_scatter_samples), unit=r"mag"))

    l_scatter_samples = hd_scatter_samples / 2.5
    lines.append(r"\newcommand{\resultScatterL}{\ensuremath{%s}}" % format_value_uncertainty(*sym_percentile(l_scatter_samples), unit=r"dex"))


    tex_path = os.path.join(write_path, "param_results.tex")
    with open(tex_path, "w") as f:
        for line in lines:
            print(line)
            f.write(line + "\n")
    print(f"Wrote result parameters LaTeX commands to {tex_path}")

compare_r = {
    "sigma_from_odds_two_sided": 3.5
}
write_results_tex_variables(df_agn, flat_samples_w0wa, cosmo_model_w0wa, compare_r, 0, 'plots/hubble')
```
